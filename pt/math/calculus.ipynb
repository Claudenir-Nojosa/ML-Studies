## <span style="color: #1e3a5f;">Quociente de Diferenças (Difference Quotient)</span>

### <span style="color: #2c5282;">Conceito Preliminar: Derivada em Uma Variável</span>

Antes de falarmos sobre derivadas parciais, vamos relembrar a derivada de uma função de **uma variável**.

Para uma função $f: \mathbb{R} \to \mathbb{R}$, a **derivada** de $f$ no ponto $x$ é definida pelo limite:

$$f'(x) = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}$$

A expressão:

$$\frac{f(x + h) - f(x)}{h}$$

é chamada de **quociente de diferenças** (ou **quociente diferencial**).

**Interpretação geométrica:** A derivada representa a **taxa de variação instantânea** da função, ou a **inclinação da reta tangente** ao gráfico de $f$ no ponto $x$.

---

## <span style="color: #1e3a5f;">Funções de Várias Variáveis</span>

### <span style="color: #2c5282;">Motivação</span>

Em muitas situações práticas, uma quantidade depende de **múltiplas variáveis**. Por exemplo:

- A temperatura $T$ em uma sala depende da posição $(x, y, z)$: $T = f(x, y, z)$
- O lucro $L$ de uma empresa depende dos preços de vários produtos: $L = g(p_1, p_2, \ldots, p_n)$
- A altura de uma montanha depende da latitude e longitude: $h = h(x, y)$

Para funções de várias variáveis, precisamos de uma noção de derivada que capture como a função varia em relação a **cada variável individualmente**.

---

### <span style="color: #2c5282;">Notação</span>

Uma função de **duas variáveis** é denotada por:

$$f: \mathbb{R}^2 \to \mathbb{R}$$
$$(x, y) \mapsto f(x, y)$$

Uma função de **$n$ variáveis**:

$$f: \mathbb{R}^n \to \mathbb{R}$$
$$(x_1, x_2, \ldots, x_n) \mapsto f(x_1, x_2, \ldots, x_n)$$

---

## <span style="color: #1e3a5f;">Derivadas Parciais</span>

### <span style="color: #2c5282;">Definição Intuitiva</span>

A **derivada parcial** de uma função de várias variáveis em relação a uma variável específica mede a **taxa de variação** da função quando **apenas aquela variável varia**, mantendo todas as outras **constantes**.

---

### <span style="color: #2c5282;">Definição Formal: Derivada Parcial em Relação a $x$</span>

Seja $f: \mathbb{R}^2 \to \mathbb{R}$ uma função de duas variáveis. A **derivada parcial** de $f$ em relação a $x$ no ponto $(x_0, y_0)$ é definida por:

$$\frac{\partial f}{\partial x}(x_0, y_0) = \lim_{h \to 0} \frac{f(x_0 + h, y_0) - f(x_0, y_0)}{h}$$

**Notações equivalentes:**

$$\frac{\partial f}{\partial x} = f_x = \partial_x f = D_x f$$

---

### <span style="color: #2c5282;">Definição Formal: Derivada Parcial em Relação a $y$</span>

A **derivada parcial** de $f$ em relação a $y$ no ponto $(x_0, y_0)$ é:

$$\frac{\partial f}{\partial y}(x_0, y_0) = \lim_{h \to 0} \frac{f(x_0, y_0 + h) - f(x_0, y_0)}{h}$$

**Notações equivalentes:**

$$\frac{\partial f}{\partial y} = f_y = \partial_y f = D_y f$$

---

### <span style="color: #2c5282;">Interpretação Geométrica</span>

**Derivada parcial em relação a $x$:**
- Mantemos $y$ fixo (constante)
- Variamos apenas $x$
- A derivada parcial $\frac{\partial f}{\partial x}$ representa a inclinação do gráfico na **direção do eixo $x$**

**Derivada parcial em relação a $y$:**
- Mantemos $x$ fixo (constante)
- Variamos apenas $y$
- A derivada parcial $\frac{\partial f}{\partial y}$ representa a inclinação do gráfico na **direção do eixo $y$**

---

### <span style="color: #2c5282;">Interpretação Prática</span>

Se $T(x, y)$ representa a temperatura em uma posição $(x, y)$:

- $\frac{\partial T}{\partial x}$ indica **quanto a temperatura aumenta** quando nos movemos na direção $x$ (leste-oeste)
- $\frac{\partial T}{\partial y}$ indica **quanto a temperatura aumenta** quando nos movemos na direção $y$ (norte-sul)

---

## <span style="color: #1e3a5f;">Cálculo de Derivadas Parciais</span>

### <span style="color: #2c5282;">Regra Prática</span>

Para calcular a derivada parcial de $f(x, y)$ em relação a uma variável:

1. **Trate todas as outras variáveis como constantes**
2. **Derive normalmente** em relação à variável desejada, usando as regras de derivação de uma variável

---

## <span style="color: #1e3a5f;">Exemplo 1: Função Polinomial Simples</span>

### <span style="color: #2c5282;">Enunciado</span>

Seja $f(x, y) = x^2 + 3xy + y^2$. Calcule $\frac{\partial f}{\partial x}$ e $\frac{\partial f}{\partial y}$.

**Solução:**

**Calculando $\frac{\partial f}{\partial x}$:**

Tratamos $y$ como **constante** e derivamos em relação a $x$:

$$\frac{\partial f}{\partial x} = \frac{\partial}{\partial x}(x^2 + 3xy + y^2)$$

- $\frac{\partial}{\partial x}(x^2) = 2x$
- $\frac{\partial}{\partial x}(3xy) = 3y$ (pois $y$ é constante, é como derivar $3y \cdot x$)
- $\frac{\partial}{\partial x}(y^2) = 0$ (constante em relação a $x$)

$$\frac{\partial f}{\partial x} = 2x + 3y$$

---

**Calculando $\frac{\partial f}{\partial y}$:**

Tratamos $x$ como **constante** e derivamos em relação a $y$:

$$\frac{\partial f}{\partial y} = \frac{\partial}{\partial y}(x^2 + 3xy + y^2)$$

- $\frac{\partial}{\partial y}(x^2) = 0$ (constante em relação a $y$)
- $\frac{\partial}{\partial y}(3xy) = 3x$ (pois $x$ é constante)
- $\frac{\partial}{\partial y}(y^2) = 2y$

$$\frac{\partial f}{\partial y} = 3x + 2y$$

**Resposta:**

$$\frac{\partial f}{\partial x} = 2x + 3y$$

$$\frac{\partial f}{\partial y} = 3x + 2y$$

---

## <span style="color: #1e3a5f;">Exemplo 2: Função Exponencial</span>

### <span style="color: #2c5282;">Enunciado</span>

Seja $g(x, y) = e^{xy}$. Calcule $\frac{\partial g}{\partial x}$ e $\frac{\partial g}{\partial y}$.

**Solução:**

**Calculando $\frac{\partial g}{\partial x}$:**

Tratamos $y$ como constante:

$$\frac{\partial g}{\partial x} = \frac{\partial}{\partial x}(e^{xy})$$

Usando a regra da cadeia (derivada de $e^u$ é $e^u \cdot u'$):

$$u = xy \Rightarrow \frac{\partial u}{\partial x} = y$$

$$\frac{\partial g}{\partial x} = e^{xy} \cdot y = ye^{xy}$$

---

**Calculando $\frac{\partial g}{\partial y}$:**

Tratamos $x$ como constante:

$$\frac{\partial g}{\partial y} = \frac{\partial}{\partial y}(e^{xy})$$

$$u = xy \Rightarrow \frac{\partial u}{\partial y} = x$$

$$\frac{\partial g}{\partial y} = e^{xy} \cdot x = xe^{xy}$$

**Resposta:**

$$\frac{\partial g}{\partial x} = ye^{xy}$$

$$\frac{\partial g}{\partial y} = xe^{xy}$$

---

## <span style="color: #1e3a5f;">Exemplo 3: Função com Produtos e Quocientes</span>

### <span style="color: #2c5282;">Enunciado</span>

Seja $h(x, y) = \frac{x^2y}{x + y}$. Calcule $\frac{\partial h}{\partial x}$ e $\frac{\partial h}{\partial y}$.

**Solução:**

**Calculando $\frac{\partial h}{\partial x}$:**

Tratamos $y$ como constante e usamos a **regra do quociente**:

$$\frac{\partial h}{\partial x} = \frac{\partial}{\partial x}\left(\frac{x^2y}{x + y}\right)$$

Regra do quociente: $\left(\frac{u}{v}\right)' = \frac{u'v - uv'}{v^2}$

Onde:
- $u = x^2y \Rightarrow \frac{\partial u}{\partial x} = 2xy$ (y é constante)
- $v = x + y \Rightarrow \frac{\partial v}{\partial x} = 1$

$$\frac{\partial h}{\partial x} = \frac{(2xy)(x + y) - (x^2y)(1)}{(x + y)^2}$$

$$= \frac{2x^2y + 2xy^2 - x^2y}{(x + y)^2}$$

$$= \frac{x^2y + 2xy^2}{(x + y)^2}$$

$$= \frac{xy(x + 2y)}{(x + y)^2}$$

---

**Calculando $\frac{\partial h}{\partial y}$:**

Tratamos $x$ como constante:

$$\frac{\partial h}{\partial y} = \frac{\partial}{\partial y}\left(\frac{x^2y}{x + y}\right)$$

Onde:
- $u = x^2y \Rightarrow \frac{\partial u}{\partial y} = x^2$ (x é constante)
- $v = x + y \Rightarrow \frac{\partial v}{\partial y} = 1$

$$\frac{\partial h}{\partial y} = \frac{(x^2)(x + y) - (x^2y)(1)}{(x + y)^2}$$

$$= \frac{x^3 + x^2y - x^2y}{(x + y)^2}$$

$$= \frac{x^3}{(x + y)^2}$$

**Resposta:**

$$\frac{\partial h}{\partial x} = \frac{xy(x + 2y)}{(x + y)^2}$$

$$\frac{\partial h}{\partial y} = \frac{x^3}{(x + y)^2}$$

---

## <span style="color: #1e3a5f;">Exemplo 4: Função de Três Variáveis</span>

### <span style="color: #2c5282;">Enunciado</span>

Seja $f(x, y, z) = x^2yz + xy^2 + z^3$. Calcule $\frac{\partial f}{\partial x}$, $\frac{\partial f}{\partial y}$ e $\frac{\partial f}{\partial z}$.

**Solução:**

**Calculando $\frac{\partial f}{\partial x}$:**

Tratamos $y$ e $z$ como constantes:

$$\frac{\partial f}{\partial x} = \frac{\partial}{\partial x}(x^2yz + xy^2 + z^3)$$

- $\frac{\partial}{\partial x}(x^2yz) = 2xyz$ (y e z são constantes)
- $\frac{\partial}{\partial x}(xy^2) = y^2$
- $\frac{\partial}{\partial x}(z^3) = 0$

$$\frac{\partial f}{\partial x} = 2xyz + y^2$$

---

**Calculando $\frac{\partial f}{\partial y}$:**

Tratamos $x$ e $z$ como constantes:

$$\frac{\partial f}{\partial y} = \frac{\partial}{\partial y}(x^2yz + xy^2 + z^3)$$

- $\frac{\partial}{\partial y}(x^2yz) = x^2z$
- $\frac{\partial}{\partial y}(xy^2) = 2xy$
- $\frac{\partial}{\partial y}(z^3) = 0$

$$\frac{\partial f}{\partial y} = x^2z + 2xy$$

---

**Calculando $\frac{\partial f}{\partial z}$:**

Tratamos $x$ e $y$ como constantes:

$$\frac{\partial f}{\partial z} = \frac{\partial}{\partial z}(x^2yz + xy^2 + z^3)$$

- $\frac{\partial}{\partial z}(x^2yz) = x^2y$
- $\frac{\partial}{\partial z}(xy^2) = 0$
- $\frac{\partial}{\partial z}(z^3) = 3z^2$

$$\frac{\partial f}{\partial z} = x^2y + 3z^2$$

**Resposta:**

$$\frac{\partial f}{\partial x} = 2xyz + y^2$$

$$\frac{\partial f}{\partial y} = x^2z + 2xy$$

$$\frac{\partial f}{\partial z} = x^2y + 3z^2$$

---

## <span style="color: #1e3a5f;">Exemplo 5: Avaliando em um Ponto Específico</span>

### <span style="color: #2c5282;">Enunciado</span>

Seja $f(x, y) = x^3 + 2xy - y^2$. Calcule $\frac{\partial f}{\partial x}(1, 2)$ e $\frac{\partial f}{\partial y}(1, 2)$.

**Solução:**

**Passo 1:** Calcular as derivadas parciais em geral:

$$\frac{\partial f}{\partial x} = 3x^2 + 2y$$

$$\frac{\partial f}{\partial y} = 2x - 2y$$

**Passo 2:** Avaliar no ponto $(1, 2)$:

$$\frac{\partial f}{\partial x}(1, 2) = 3(1)^2 + 2(2) = 3 + 4 = 7$$

$$\frac{\partial f}{\partial y}(1, 2) = 2(1) - 2(2) = 2 - 4 = -2$$

**Resposta:**

No ponto $(1, 2)$:
- A taxa de variação na direção $x$ é $7$
- A taxa de variação na direção $y$ é $-2$

**Interpretação:** Se nos movermos na direção $x$ mantendo $y = 2$, a função aumenta a uma taxa de 7 unidades por unidade de $x$. Se nos movermos na direção $y$ mantendo $x = 1$, a função diminui a uma taxa de 2 unidades por unidade de $y$.

---

## <span style="color: #1e3a5f;">Derivadas Parciais de Ordem Superior</span>

### <span style="color: #2c5282;">Definição</span>

Podemos calcular derivadas parciais de derivadas parciais, obtendo **derivadas parciais de segunda ordem** (e ordens superiores).

**Notações:**

$$\frac{\partial^2 f}{\partial x^2} = f_{xx} = \text{derivar duas vezes em relação a } x$$

$$\frac{\partial^2 f}{\partial y^2} = f_{yy} = \text{derivar duas vezes em relação a } y$$

$$\frac{\partial^2 f}{\partial x \partial y} = f_{xy} = \text{derivar primeiro em relação a } y, \text{ depois em relação a } x$$

$$\frac{\partial^2 f}{\partial y \partial x} = f_{yx} = \text{derivar primeiro em relação a } x, \text{ depois em relação a } y$$

---

### <span style="color: #2c5282;">Teorema de Schwarz (Clairaut)</span>

Se as derivadas parciais mistas $\frac{\partial^2 f}{\partial x \partial y}$ e $\frac{\partial^2 f}{\partial y \partial x}$ são **contínuas**, então:

$$\frac{\partial^2 f}{\partial x \partial y} = \frac{\partial^2 f}{\partial y \partial x}$$

Ou seja, $f_{xy} = f_{yx}$ (a ordem de derivação não importa).

---

### <span style="color: #2c5282;">Exemplo 6: Derivadas de Segunda Ordem</span>

Seja $f(x, y) = x^3y^2 + 2xy$. Calcule todas as derivadas parciais de segunda ordem.

**Solução:**

**Passo 1:** Derivadas de primeira ordem:

$$\frac{\partial f}{\partial x} = 3x^2y^2 + 2y$$

$$\frac{\partial f}{\partial y} = 2x^3y + 2x$$

---

**Passo 2:** Derivadas de segunda ordem:

$$\frac{\partial^2 f}{\partial x^2} = \frac{\partial}{\partial x}(3x^2y^2 + 2y) = 6xy^2$$

$$\frac{\partial^2 f}{\partial y^2} = \frac{\partial}{\partial y}(2x^3y + 2x) = 2x^3$$

$$\frac{\partial^2 f}{\partial x \partial y} = \frac{\partial}{\partial x}(2x^3y + 2x) = 6x^2y + 2$$

$$\frac{\partial^2 f}{\partial y \partial x} = \frac{\partial}{\partial y}(3x^2y^2 + 2y) = 6x^2y + 2$$

**Verificação do Teorema de Schwarz:**

$$f_{xy} = 6x^2y + 2 = f_{yx}$$

---

## <span style="color: #1e3a5f;">Resumo</span>

### <span style="color: #2c5282;">Conceitos Principais</span>

- **Quociente de diferenças:** $\frac{f(x + h) - f(x)}{h}$, base da definição de derivada

- **Derivada parcial:** Taxa de variação quando apenas uma variável muda

- **Notação:** $\frac{\partial f}{\partial x}$, $f_x$, $\partial_x f$

- **Regra prática:** Trate as outras variáveis como constantes e derive normalmente

- **Interpretação:** Inclinação do gráfico na direção de cada eixo

- **Derivadas mistas:** Ordem não importa (teorema de Schwarz)

## <span style="color: #1e3a5f;">Jacobiano e Hessiana</span>

### <span style="color: #2c5282;">Motivação</span>

Em funções de **várias variáveis**, não estamos interessados apenas em como a função varia em relação a **uma variável por vez**, mas também:

- Como **várias saídas** variam simultaneamente em relação a várias entradas
- Como a função **curva**, isto é, como suas derivadas parciais **mudam**

Para isso, introduzimos dois objetos fundamentais no cálculo vetorial e na ciência de dados:

- **Matriz Jacobiana (Jacobian)**
- **Matriz Hessiana (Hessian)**

Esses conceitos são centrais em:
- Otimização
- Machine Learning
- Redes neurais
- Métodos numéricos
- Análise de sensibilidade

---

## <span style="color: #1e3a5f;">O Jacobiano</span>

### <span style="color: #2c5282;">Definição Intuitiva</span>

O **Jacobiano** generaliza a derivada para funções que possuem:

- **Várias variáveis de entrada**
- **Uma ou várias variáveis de saída**

Ele descreve como **pequenas variações na entrada** afetam a **saída** da função.

Em outras palavras, o Jacobiano representa a **melhor aproximação linear local** de uma função multivariada.

---

### <span style="color: #2c5282;">Função Vetorial</span>

Considere uma função:

$$
\mathbf{f}: \mathbb{R}^n \to \mathbb{R}^m
$$

Definida por:

$$
\mathbf{f}(x_1, x_2, \ldots, x_n)
=
\begin{bmatrix}
f_1(x_1, \ldots, x_n) \\
f_2(x_1, \ldots, x_n) \\
\vdots \\
f_m(x_1, \ldots, x_n)
\end{bmatrix}
$$

Cada $f_i$ é uma função escalar de $n$ variáveis.

---

### <span style="color: #2c5282;">Definição Formal: Matriz Jacobiana</span>

A **matriz Jacobiana** de $\mathbf{f}$ é a matriz $m \times n$:

$$
J_{\mathbf{f}} =
\begin{bmatrix}
\frac{\partial f_1}{\partial x_1} & \frac{\partial f_1}{\partial x_2} & \cdots & \frac{\partial f_1}{\partial x_n} \\
\frac{\partial f_2}{\partial x_1} & \frac{\partial f_2}{\partial x_2} & \cdots & \frac{\partial f_2}{\partial x_n} \\
\vdots & \vdots & \ddots & \vdots \\
\frac{\partial f_m}{\partial x_1} & \frac{\partial f_m}{\partial x_2} & \cdots & \frac{\partial f_m}{\partial x_n}
\end{bmatrix}
$$

---

### <span style="color: #2c5282;">Caso Especial: Função Escalar</span>

Se $f: \mathbb{R}^n \to \mathbb{R}$ (uma única saída), o Jacobiano reduz-se a um **vetor linha**:

$$
J_f =
\begin{bmatrix}
\frac{\partial f}{\partial x_1} &
\frac{\partial f}{\partial x_2} &
\cdots &
\frac{\partial f}{\partial x_n}
\end{bmatrix}
$$

Esse vetor é exatamente o **gradiente transposto**:

$$
J_f = \nabla f^\top
$$

---

### <span style="color: #2c5282;">Exemplo 1: Jacobiano de Função Vetorial</span>

Considere:

$$
\mathbf{f}(x, y) =
\begin{bmatrix}
x^2 + y \\
xy
\end{bmatrix}
$$

Calculamos as derivadas parciais:

- $f_1(x, y) = x^2 + y$
- $f_2(x, y) = xy$

Jacobiano:

$$
J_{\mathbf{f}} =
\begin{bmatrix}
\frac{\partial f_1}{\partial x} & \frac{\partial f_1}{\partial y} \\
\frac{\partial f_2}{\partial x} & \frac{\partial f_2}{\partial y}
\end{bmatrix}
=
\begin{bmatrix}
2x & 1 \\
y & x
\end{bmatrix}
$$

---

### <span style="color: #2c5282;">Interpretação Geométrica</span>

- Cada **linha**: variação de uma componente da saída
- Cada **coluna**: sensibilidade da função a uma variável de entrada
- O Jacobiano aproxima a função por uma transformação linear local

---

## <span style="color: #1e3a5f;">A Hessiana</span>

### <span style="color: #2c5282;">Definição Intuitiva</span>

A **Hessiana** captura a **curvatura** de uma função escalar de várias variáveis.

Enquanto o **Jacobiano** fala de **inclinação**, a **Hessiana** fala de:

- Concavidade
- Convexidade
- Pontos de mínimo e máximo
- Comportamento local da função

Ela é fundamental em **otimização de segunda ordem**.

---

### <span style="color: #2c5282;">Definição Formal: Matriz Hessiana</span>

Seja $f: \mathbb{R}^n \to \mathbb{R}$ uma função escalar.

A **matriz Hessiana** é a matriz $n \times n$ das derivadas parciais de segunda ordem:

$$
H_f =
\begin{bmatrix}
\frac{\partial^2 f}{\partial x_1^2} &
\frac{\partial^2 f}{\partial x_1 \partial x_2} &
\cdots &
\frac{\partial^2 f}{\partial x_1 \partial x_n} \\

\frac{\partial^2 f}{\partial x_2 \partial x_1} &
\frac{\partial^2 f}{\partial x_2^2} &
\cdots &
\frac{\partial^2 f}{\partial x_2 \partial x_n} \\

\vdots & \vdots & \ddots & \vdots \\

\frac{\partial^2 f}{\partial x_n \partial x_1} &
\frac{\partial^2 f}{\partial x_n \partial x_2} &
\cdots &
\frac{\partial^2 f}{\partial x_n^2}
\end{bmatrix}
$$

---

### <span style="color: #2c5282;">Propriedade Importante</span>

Se as derivadas de segunda ordem são **contínuas**, então a Hessiana é **simétrica**:

$$
\frac{\partial^2 f}{\partial x_i \partial x_j}
=
\frac{\partial^2 f}{\partial x_j \partial x_i}
$$

Isso é consequência direta do **Teorema de Schwarz (Clairaut)**.

---

### <span style="color: #2c5282;">Exemplo 2: Hessiana em Duas Variáveis</span>

Considere:

$$
f(x, y) = x^2 + 3xy + y^2
$$

Derivadas de primeira ordem:

$$
f_x = 2x + 3y
\quad
f_y = 3x + 2y
$$

Derivadas de segunda ordem:

$$
f_{xx} = 2
\quad
f_{yy} = 2
\quad
f_{xy} = f_{yx} = 3
$$

Hessiana:

$$
H_f =
\begin{bmatrix}
2 & 3 \\
3 & 2
\end{bmatrix}
$$

---

### <span style="color: #2c5282;">Interpretação em Otimização</span>

A Hessiana permite classificar pontos críticos:

- **Definida positiva** → mínimo local
- **Definida negativa** → máximo local
- **Indefinida** → ponto de sela

Esses conceitos são essenciais em:
- Gradient Descent
- Newton's Method
- Treinamento de modelos de ML

---

## <span style="color: #1e3a5f;">Resumo</span>

### <span style="color: #2c5282;">Comparação Jacobiano vs Hessiana</span>

| Conceito | Jacobiano | Hessiana |
|--------|-----------|----------|
| Ordem das derivadas | Primeira | Segunda |
| Tipo de função | Vetorial ou escalar | Escalar |
| Dimensão | $m \times n$ | $n \times n$ |
| Interpretação | Inclinação / sensibilidade | Curvatura |
| Uso principal | Aproximação linear | Otimização |

---

### <span style="color: #2c5282;">Ideia Central</span>

- **Jacobiano:** como a função muda
- **Hessiana:** como essa mudança muda

Ambos são pilares matemáticos da **ciência de dados moderna**.

---


## <span style="color: #1e3a5f;">Série de Taylor (Taylor Series)</span>

### <span style="color: #2c5282;">Motivação</span>

Em muitos problemas de matemática aplicada, ciência de dados e machine learning, não precisamos conhecer uma função **exatamente**, mas sim:

- Uma **boa aproximação local**
- Um modelo simples que se comporte bem **perto de um ponto**
- Uma forma de transformar funções complexas em **polinômios**

A **Série de Taylor** fornece exatamente isso:  
uma forma sistemática de **aproximar funções usando derivadas**.

---

## <span style="color: #1e3a5f;">Taylor em Uma Variável</span>

### <span style="color: #2c5282;">Ideia Intuitiva</span>

Dada uma função $f(x)$ suficientemente suave, podemos aproximá-la perto de um ponto $a$ usando um **polinômio** construído a partir de suas derivadas em $a$.

Cada termo adiciona **mais precisão** à aproximação.

---

### <span style="color: #2c5282;">Expansão de Taylor</span>

A **Série de Taylor** de $f(x)$ em torno do ponto $a$ é:

$$
f(x) =
f(a)
+ f'(a)(x - a)
+ \frac{f''(a)}{2!}(x - a)^2
+ \frac{f^{(3)}(a)}{3!}(x - a)^3
+ \cdots
$$

De forma compacta:

$$
f(x) = \sum_{k=0}^{\infty}
\frac{f^{(k)}(a)}{k!}(x - a)^k
$$

---

### <span style="color: #2c5282;">Casos Especiais</span>

- **Polinômio de Taylor de ordem 1** → Aproximação linear  
- **Polinômio de Taylor de ordem 2** → Aproximação quadrática  
- **Polinômio de Taylor de ordem $n$** → Aproximação de ordem $n$

---

### <span style="color: #2c5282;">Exemplo 1: Taylor de Ordem 1</span>

Considere:

$$
f(x) = e^x
$$

Derivadas em $a = 0$:

$$
f(0) = 1
\quad
f'(0) = 1
$$

Aproximação linear:

$$
e^x \approx 1 + x
$$

Essa é a melhor aproximação linear de $e^x$ perto de $0$.

---

### <span style="color: #2c5282;">Exemplo 2: Taylor de Ordem 2</span>

Considere novamente $f(x) = e^x$.

$$
f''(x) = e^x \Rightarrow f''(0) = 1
$$

Aproximação quadrática:

$$
e^x \approx 1 + x + \frac{x^2}{2}
$$

---

## <span style="color: #1e3a5f;">Série de Taylor em Várias Variáveis</span>

### <span style="color: #2c5282;">Generalização</span>

Para funções de várias variáveis, a ideia é a mesma:

> aproximar a função perto de um ponto usando derivadas parciais

Aqui entram diretamente:
- **Gradiente**
- **Jacobiano**
- **Hessiana**

---

### <span style="color: #2c5282;">Taylor de Primeira Ordem</span>

Seja $f: \mathbb{R}^n \to \mathbb{R}$ e um ponto $\mathbf{a}$.

A aproximação de **primeira ordem** é:

$$
f(\mathbf{x}) \approx
f(\mathbf{a})
+ \nabla f(\mathbf{a})^\top (\mathbf{x} - \mathbf{a})
$$

**Interpretação:**  
Essa é a **melhor aproximação linear local** de $f$.

---

### <span style="color: #2c5282;">Taylor de Segunda Ordem</span>

A aproximação de **segunda ordem** incorpora a curvatura da função:

$$
f(\mathbf{x}) \approx
f(\mathbf{a})
+ \nabla f(\mathbf{a})^\top (\mathbf{x} - \mathbf{a})
+ \frac{1}{2}(\mathbf{x} - \mathbf{a})^\top
H_f(\mathbf{a})
(\mathbf{x} - \mathbf{a})
$$

Onde:
- $\nabla f$ é o **gradiente**
- $H_f$ é a **matriz Hessiana**

---

### <span style="color: #2c5282;">Conexão Direta</span>

| Ordem | Objeto Matemático | Informação |
|-----|------------------|------------|
| 0 | $f(a)$ | Valor |
| 1 | Gradiente / Jacobiano | Inclinação |
| 2 | Hessiana | Curvatura |

---

## <span style="color: #1e3a5f;">Exemplo em Duas Variáveis</span>

### <span style="color: #2c5282;">Exemplo 3</span>

Considere:

$$
f(x, y) = x^2 + 3xy + y^2
$$

Expansão em torno de $(0, 0)$.

Valor da função:

$$
f(0, 0) = 0
$$

Gradiente:

$$
\nabla f =
\begin{bmatrix}
2x + 3y \\
3x + 2y
\end{bmatrix}
\Rightarrow
\nabla f(0, 0) =
\begin{bmatrix}
0 \\
0
\end{bmatrix}
$$

Hessiana:

$$
H_f =
\begin{bmatrix}
2 & 3 \\
3 & 2
\end{bmatrix}
$$

Aproximação de segunda ordem:

$$
f(x, y) \approx
\frac{1}{2}
\begin{bmatrix}
x & y
\end{bmatrix}
\begin{bmatrix}
2 & 3 \\
3 & 2
\end{bmatrix}
\begin{bmatrix}
x \\
y
\end{bmatrix}
$$

---

## <span style="color: #1e3a5f;">Importância em Machine Learning</span>

### <span style="color: #2c5282;">Por que isso importa?</span>

- **Gradient Descent** usa Taylor de primeira ordem
- **Newton's Method** usa Taylor de segunda ordem
- Funções de perda são analisadas localmente via Taylor
- A Hessiana determina a **geometria da superfície de erro**

Treinar um modelo é, essencialmente, **navegar pela série de Taylor da loss function**.

---

## <span style="color: #1e3a5f;">Resumo</span>

### <span style="color: #2c5282;">Ideias-Chave</span>

- A **Série de Taylor** aproxima funções por polinômios
- Em várias variáveis:
  - Ordem 1 → Gradiente / Jacobiano
  - Ordem 2 → Hessiana
- Taylor conecta **derivadas** a **otimização**
- É um dos pilares matemáticos do **Machine Learning**

---

### <span style="color: #2c5282;">Mensagem Central</span>

> **Gradiente diz para onde ir.  
> Hessiana diz como a superfície se curva.  
> Taylor explica por quê.**
